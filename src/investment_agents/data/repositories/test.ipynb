```python
from investment_agents.data.clients.finmind import FinMindClient
from investment_agents.data.repositories.market_regime import MarketRegimeRepository

client = FinMindClient()

regime_repository = MarketRegimeRepository(
    client
)

raw = regime_repository.get_data(
    start_date="2026-07-01",
    end_date="2026-09-18",
)

us = raw["us_market"].sort_values("date").copy()

us["spy_ret_20d_manual"] = (
    us["spy"] / us["spy"].shift(20) - 1
) * 100

print(
    us[
        ["date", "spy", "spy_ret_20d_manual"]
    ].tail(25).to_string(index=False)
)
```
